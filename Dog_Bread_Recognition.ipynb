##**<u>[2] Brief Literature Review</u>**
According to <b>Z. -Q. Zhao, et al. 2019</b> Object Recognition can be described as the problem of detecting where the object is located on the image space, and how this object can be categorized. These problems are called Object Localization and Object Classification.

###**(a)	Traditional Object Recognition**

To understand object recognition its important to talk about traditional object recognition. Traditional Object Recognition methods can be divide into 3 sections: informative region selection, feature extraction and classification. This division of traditional object recognition methods can be observed in <b> B. R. Solunke and S. R. Gengaje, 2023</b> were the authors go over these steps and establish while its possible to create a successful approach to object detection it does require a costly amount of computational power, and has a difficult time of creating a reliable function manually, in the feature extraction step, due to image diversity. These issues are also mentioned by <b> Z. -Q. Zhao, et al. 2019 </b> regarding traditional object recognition methods. <i>[Z. -Q. Zhao, et al. 2019, B. R. Solunke and S. R. Gengaje, 2023]</i>

###**(b)	Deep Learning and Object Recognition**

Object recognition using deep learning is a much more rewarding approach then traditional object recognition, this is due to the introduction of Convolutional Neural Networks (CNNs) models which can learn complicated and powerful features and learn hierarchical representations, has a deeper architecture compared to traditional methods and importantly CNNs allow to learn informative object representation automatically, reducing the need for the manual design of features. It consists of layers like the convolutional, pooling, and fully connected layers. The convolutional layer combines the input image with kernels or filters which are learned through the training process. These kernels or filters only take up a fraction of the input image.

![convolutional layers](https://ars.els-cdn.com/content/image/3-s2.0-B978012822822700003X-f03-06-9780128228227.jpg)

<i>Figure 1: Convolution process</i>

The pooling layer reduces the dimensions of the feature maps generated by the convolutional layer (e.g.: Max Pooling).

![convolutional layers](https://www.researchgate.net/publication/333593451/figure/fig2/AS:765890261966848@1559613876098/llustration-of-Max-Pooling-and-Average-Pooling-Figure-2-above-shows-an-example-of-max.png)

<i>Figure 2: Pooling process</i>

The fully connected layers use the output of the convolutional and/or pooling layers. The fully connected layers flatten the data and process it.
<i>[S. Shajun Nisha, M. Nagoor Meeral, 2021, Sakib Mostafa, Fang-Xiang Wu, 2021, S. Boopathi, U. K. Kanike, 2023, B. R. Solunke and S. R. Gengaje, 2023, Z. -Q. Zhao, et al. 2019, Bansal, M., et al, 2021]</i>

Other variations of <b>CNN</b> are: <b>R-CNN, Fast R-CNN, Faster R-CNN,</b> and <b>Mask R-CNN</b> which use region proposal based methods. Another method is regression/classification based which includes <b>YOLO, MultiBox, G-CNN</b> etc. <i>[Z. -Q. Zhao, et al. 2019]</i>

###Importing Dataset

```python
# IMPORTS
import numpy as np
import tensorflow_datasets as tfds
import tensorflow as tf
import matplotlib.pyplot as plt
import keras
from keras import layers
from keras.applications import EfficientNetB0

# VARIABLES
IMG_SIZE = 224
BATCH_SIZE = 64
DATASET = "stanford_dogs"
NUM_CLASSES = 120

#--(CHECKS AVAILABILITY AND LOADS IT)-------------------------------------------
(train_data, test_data), ds_info = tfds.load(DATASET,
                            split=['train', 'test'],
                            as_supervised=True,
                            shuffle_files=True,
                            with_info=True)
```

###Data Processing

```python
# RESIZING IMAGE
size = (IMG_SIZE, IMG_SIZE)
train_data = train_data.map(lambda image, label: (tf.image.resize(image, size), label))
test_data = test_data.map(lambda image, label: (tf.image.resize(image, size), label))
```

```python
# VISUALIZING AMOUNT OF DATA
print('Number of images in training: ' + str(len(train_data)))
print('Number of images in testing: ' + str(len(test_data)))
```

```python
# FORMATING LABELS FOR IMAGES
def format_label(label):
    string_label = label_info.int2str(label)
    return string_label.split("-")[1]

# VISUALIZING IMAGES FROM DATASET WITH LABELS
label_info = ds_info.features["label"]
for i, (image, label) in enumerate(train_data.take(9)):
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(image.numpy().astype("uint8"))
    plt.title("{}".format(format_label(label).replace("_", " ")))
    plt.axis("off")
```

###Data Augmentation

```python
# DEFINING IMAGE AUGMENTATIONS
img_augmentation_layers = [
    layers.RandomRotation(factor=0.15),
    layers.RandomTranslation(height_factor=0.1, width_factor=0.1),
    layers.RandomFlip(),
]

# FUNCTION TO AUGMENT IMAGES
def img_augmentation(images):
    for layer in img_augmentation_layers:
        images = layer(images)
    return images
```

```python
# VISUALIZING AUGMENTED IMAGES
for image, label in train_data.take(1):
    for i in range(9):
        ax = plt.subplot(3, 3, i + 1)
        aug_img = img_augmentation(np.expand_dims(image.numpy(), axis=0))
        aug_img = np.array(aug_img)
        plt.imshow(aug_img[0].astype("uint8"))
        plt.title("{}".format(format_label(label).replace("_", " ")))
        plt.axis("off")
```

###Data Pipeline and Model Building

```python
# ONE-HOT ENCODING AND AUGMENTING IMAGES FOR TRAINING DATA
def input_preprocess_train(image, label):
    image = img_augmentation(image)
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])  # This line is crucial
    label = tf.one_hot(label, NUM_CLASSES)
    return image, label

# ONE-HOT ENCODING FOR TEST DATA
def input_preprocess_test(image, label):
    label = tf.one_hot(label, NUM_CLASSES)
    return image, label


train_data = train_data.map(input_preprocess_train, num_parallel_calls=tf.data.AUTOTUNE)
train_data = train_data.batch(batch_size=BATCH_SIZE, drop_remainder=True)
train_data = train_data.prefetch(tf.data.AUTOTUNE)

test_data = test_data.map(input_preprocess_test, num_parallel_calls=tf.data.AUTOTUNE)
test_data = test_data.batch(batch_size=BATCH_SIZE, drop_remainder=True)
```

```python
# FUNCTION TO BUILD THE MODEL
def build_model(num_classes):
    # SETTING INPUTS TO THE PREDEFINED IMAGE SIZE
    inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
    # ESTABLISHING THE MODEL (EfficientNetB0)
    model = EfficientNetB0(include_top=False, input_tensor=inputs, weights="imagenet")

    # FREEZING PRETRAINED MODELS
    model.trainable = False

    # REBUILD TOP
    x = layers.GlobalAveragePooling2D(name="avg_pool")(model.output)
    x = layers.BatchNormalization()(x)

    top_dropout_rate = 0.2
    x = layers.Dropout(top_dropout_rate, name="top_dropout")(x)
    outputs = layers.Dense(num_classes, activation="softmax", name="pred")(x)

    # DEFINING LEARNING RATE
    lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
       initial_learning_rate=0.0001,
       decay_steps=5000,
       decay_rate=0.9
   )

    # COMPILE MODEL
    model = keras.Model(inputs, outputs, name="EfficientNet")
    model.compile(
        loss='categorical_crossentropy',
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr_schedule),
        metrics=['accuracy']
    )
    return model
```

###Model Training and Evaluation

```python
# ESTABLISHING THE MODEL
model = build_model(num_classes=NUM_CLASSES)

# SETTING THE EPOCHS AND TRAINING THE MODEL
epochs = 10
history = model.fit(train_data, epochs=epochs, validation_data=test_data)
```

```python
# EVALUATING MODEL ACCURACY
loss, accuracy = model.evaluate(test_data)
print('\nTest accuracy: ', accuracy, '\nIn percentage: {:.2f}%'.format(accuracy * 100))
```

```python
# VISUALIZING MODEL
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

loss = history.history['loss']
val_loss = history.history['val_loss']

epochs_range = range(epochs)

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.show()
```

###Predictions and Confusion Matrix

```python
probability_model = tf.keras.Sequential([model,
                                         tf.keras.layers.Softmax()])
probabilities = probability_model.predict(test_data)
print('Size of variable probabilities: ' + str(probabilities.shape))
print('Probabilities:')
print(probabilities)
```

```python
# RECIEVES MAXIMUM PROBABILITY OF EACH IMAGE
predictions = np.argmax(probabilities, axis = 1)

print('Size of variable predictions: ' + str(predictions.shape))
print('Predictions:')
print(predictions)
```

```python
test_labels = []
for images, labels in test_data:
    labels = tf.argmax(labels, axis=1).numpy()
    test_labels.extend(labels)

test_labels = np.array(test_labels)
```

```python
import seaborn as sns

# STORING CLASS NAMES
class_names = [ds_info.features["label"].int2str(i).split("-")[1] for i in range(ds_info.features["label"].num_classes)]

def plot_confusion_matrix(actual, predicted, labels):
  cm = tf.math.confusion_matrix(actual, predicted)
  # FIG SIZE
  fig, ax = plt.subplots(figsize=(80, 80))

  # HEATMAP
  sns.heatmap(cm, annot=True, fmt='g', ax=ax)

  ax.set_title('Confusion matrix')
  ax.set_xlabel('Prediction')
  ax.set_ylabel('Label')

  # GET NUM OF UNIQUE CLASSES
  num_classes = cm.shape[0]

  # SETTING TICK LAYERS BASED ON UNIQUE CLASSES
  tick_labels = labels[:num_classes]

  #ROTATE X-AXIS AND SETTING FONT SIZE
  ax.set_xticklabels(tick_labels, rotation=90, ha='right', fontsize=15) # Use tick_labels here
  ax.set_yticklabels(tick_labels, rotation=0, fontsize=15) # Use tick_labels here

  # SETTING LAYOUT TO PREVENT OVERLAPPING
  plt.tight_layout()

  plt.show()

plot_confusion_matrix(test_labels, predictions, class_names)
```
